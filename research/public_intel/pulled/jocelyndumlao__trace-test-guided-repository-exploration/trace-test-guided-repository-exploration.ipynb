```python
import json
import urllib.request
import pandas as pd
from IPython.display import display

REPORT_URL = (
    "https://raw.githubusercontent.com/"
    "jcdumlao14/gemma-4-developer-agent-paper-track/"
    "main/experiments/results/"
    "TRACE_GEMMA4_CONSOLIDATED_PILOT_REPORT.json"
)

request = urllib.request.Request(
    REPORT_URL,
    headers={"User-Agent": "TRACE-Kaggle-Notebook/1.0"}
)

with urllib.request.urlopen(request, timeout=30) as response:
    report = json.loads(response.read().decode("utf-8-sig"))

cases = report["cases"]

assert len(cases) == 4, f"Expected four cases, found {len(cases)}"

print("TRACE pilot report loaded successfully.")
print("Project:", report.get("project"))
print("Number of cases:", len(cases))

display(pd.DataFrame([
    {
        "Case": case["id"],
        "Candidate provenance": case.get("provenance", "Not recorded"),
        "Target passed": case.get("target_passed"),
        "Controls": case.get("controls_passed", "Not separately recorded"),
        "Broader validation": case.get("broader_validation", "Not recorded"),
    }
    for case in cases
]))
```

```python
# TRACE | Publication-style target outcome chart

import matplotlib.pyplot as plt
from matplotlib.patches import Patch

trace_header(
    "02 | Recorded Target-Test Outcomes",
    "Historical report entries — not fresh test executions"
)

labels = [c["id"] for c in cases]
values = [
    1 if c.get("target_passed") is True else 0
    for c in cases
]

# Distinguish projects visually, not candidate success attribution.
bar_colors = [
    TRACE_BLUE if name.startswith("sqlfluff")
    else TRACE_PURPLE
    for name in labels
]

fig, ax = plt.subplots(figsize=(11, 5.2))
bars = ax.barh(
    labels,
    values,
    color=bar_colors,
    height=0.58
)

ax.invert_yaxis()
ax.set_xlim(0, 1.35)
ax.set_xticks([0, 1])
ax.set_xticklabels(["Not recorded as passed", "Recorded as passed"])
ax.set_xlabel("Target-test status in consolidated report", labelpad=12)
ax.set_title(
    "TRACE Four-Case Pilot: Recorded Target Outcomes",
    fontsize=15,
    fontweight="bold",
    color=TRACE_TEXT,
    pad=18
)

ax.spines[["top", "right", "left"]].set_visible(False)
ax.grid(axis="x", linestyle="--", alpha=0.2)
ax.set_axisbelow(True)
ax.tick_params(axis="y", length=0, labelsize=10)

for bar in bars:
    passed = bar.get_width() == 1
    ax.text(
        bar.get_width() + 0.035,
        bar.get_y() + bar.get_height() / 2,
        "PASS*" if passed else "CHECK",
        va="center",
        fontsize=10,
        fontweight="bold",
        color=TRACE_GREEN if passed else TRACE_AMBER
    )

ax.legend(
    handles=[
        Patch(facecolor=TRACE_BLUE, label="SQLFluff"),
        Patch(facecolor=TRACE_PURPLE, label="Marshmallow")
    ],
    loc="lower right",
    frameon=False
)

fig.text(
    0.01, -0.015,
    "*Status is taken from the recorded report. "
    "Candidate provenance differs across cases.",
    fontsize=9,
    color=TRACE_MUTED
)

plt.tight_layout()
plt.show()

trace_note(
    "<b>Interpretation:</b> The plot describes recorded target-test outcomes. "
    "It is not an autonomous repair success rate for Gemma or TRACE."
)
```
